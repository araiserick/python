Задание 1. Получение данных из публичного API

 

Выберите публичный API. Например, JSONPlaceholder.

Напишите скрипт, который:

отправляет GET-запрос к /posts,

извлекает и выводит заголовки и тела первых 5 постов.

 

```python
import requests

# 1. Указываем URL публичного API
url = "https://jsonplaceholder.typicode.com/posts"

# 2. Отправляем GET-запрос
response = requests.get(url)

# 3. Проверяем, что запрос прошел успешно (код 200)
if response.status_code == 200:
    # Преобразуем ответ из формата JSON в список словарей Python
    posts = response.json()
    
    # 4. Берем срез (slice) первых 5 элементов списка
    first_five_posts = posts[:5]
    
    # 5. Извлекаем и выводим заголовки и тела
    print(f"Успешно получено {len(first_five_posts)} постов:\n")
    print("=" * 50)
    
    for i, post in enumerate(first_five_posts, start=1):
        print(f"Пост №{post['id']}")
        print(f"Заголовок: {post['title']}")
        print(f"Тело:\n{post['body']}")
        print("-" * 50)
else:
    print(f"Ошибка при запросе. Код состояния: {response.status_code}")

```

Задание 2. Работа с параметрами запроса

 

Используйте API OpenWeather для получения данных о погоде.

Напишите программу, которая:

принимает название города от пользователя,

отправляет GET-запрос к API и выводит текущую температуру и описание погоды.

```python
# выполнил согласно документации API
import requests

# Вставьте сюда ваш API ключ
API_KEY = "ВАШ_API_КЛЮЧ_ЗДЕСЬ"

# 1. Принимаем название города от пользователя
city = input("Введите название города (например, Moscow): ")

# 2. Формируем параметры запроса
url = "https://api.openweathermap.org/data/2.5/weather"
params = {
    "q": city,
    "appid": API_KEY,
    "units": "metric",  # Температура в Цельсиях (по умолчанию в Кельвинах)
    "lang": "ru"        # Описание погоды на русском языке
}

# 3. Отправляем GET-запрос
response = requests.get(url, params=params)

# 4. Обрабатываем ответ
if response.status_code == 200:
    data = response.json()
    
    # Извлекаем нужные данные из JSON
    temperature = data["main"]["temp"]
    description = data["weather"][0]["description"]
    feels_like = data["main"]["feels_like"]
    
    print(f"\n🌤  Погода в городе {data['name']}:")
    print(f"   Температура: {temperature}°C (ощущается как {feels_like}°C)")
    print(f"   Описание: {description.capitalize()}")
else:
    # Обработка ошибок (неверный город, неверный ключ и т.д.)
    error_msg = response.json().get("message", "Неизвестная ошибка")
    print(f"\n❌ Ошибка {response.status_code}: {error_msg}")
```
