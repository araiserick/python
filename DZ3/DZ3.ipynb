### Задание 1

Переведите содержимое файла purchase_log.txt в словарь purchases вида:
 {'1840e0b9d4': 'Продукты', ...}

Пример работы программы при выводе первых двух элементов словаря purchases:

1840e0b9d4 ‘Продукты‘

4e4f90fcfb  ‘Электроника‘

```python
purchases = {}

with open('purchase_log.txt', encoding='utf-8') as file:
    for line in file:
        line = line.strip()
        if not line:
            continue

        fields = [field.strip() for field in line.split(';')]

        purchase_id = fields[0]
        category = fields[-1]

        purchases[purchase_id] = category

for purchase_id, category in list(purchases.items())[:2]:
    print(purchase_id, category)
```

### Задание 2

Для каждого user_id в файле visit_log.csv определите третий столбец с категорией покупки. Если покупка была, сам файл visit_log.csv изменять не надо.

Запишите в файл funnel.csv визиты из файла visit_log.csv, в которых были покупки с указанием категории.

Учтите условия для данных:

содержимое purchase_log.txt помещается в оперативную память компьютера;

содержимое visit_log.csv — нет, используйте только построчную обработку этого файла.

 

Примеры работы программы при выводе первых трёх строк файла funnel.csv:

user_id,source,category

1840e0b9d4,other,Продукты

4e4f90fcfb,context,Электроника



```python
import csv

purchases = {}
with open('purchase_log.txt', encoding='utf-8') as f:
    for line in f:
        line = line.strip()
        if not line:
            continue
        parts = [p.strip() for p in line.split(';')]
        purchases[parts[0]] = parts[-1]

# 2. Построчно обрабатываем visit_log.csv и пишем funnel.csv
with open('visit_log.csv', encoding='utf-8', newline='') as fin, \
     open('funnel.csv', 'w', encoding='utf-8', newline='') as fout:

    reader = csv.reader(fin)
    writer = csv.writer(fout)

    # Пропускаем заголовок исходного файла
    next(reader)
    # Записываем заголовок результата
    writer.writerow(['user_id', 'source', 'category'])

    # Построчная обработка — в памяти держим только одну строку
    for row in reader:
        user_id = row[0].strip()
        source = row[1].strip()

        if user_id in purchases:
            writer.writerow([user_id, source, purchases[user_id]])
```
